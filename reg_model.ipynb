```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('placement.csv')
df.head()
```

```python
X = df[['cgpa']]
y = df['package']
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=20)
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
lr = LinearRegression()
model = lr.fit(X_train,y_train)
```

```python
predict = model.predict(X_test)
```

```python
print(model.predict([[9]]))
```

```python
#Evalute the model 
from sklearn.metrics import r2_score
score  = r2_score(y_test,predict)
score
```

```python
# Mean squred Error
from sklearn.metrics import mean_squared_error
mean_squared = mean_squared_error(y_test,predict)
mean_squared
```

```python
# Mean Absolute Error
from sklearn.metrics import mean_absolute_error
mean_absolute = mean_absolute_error(y_test, predict)
mean_absolute
```

```python
# Root mean Squred Error
from sklearn.metrics import root_mean_squared_error
root_mean = root_mean_squared_error(y_test,predict)
root_mean
```

RMSE,MAE, MSE =  If the value is low model is good 
r2 = if the value is model is good

```python
# use cross-Validation insted of mae,mse, r2
from sklearn.model_selection import cross_val_score
score = cross_val_score(model,X,y, cv=5,scoring='r2')
score.mean()
```
